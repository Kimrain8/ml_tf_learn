```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn import datasets
import matplotlib.pylab as pyb
```

```python
# 准备数据集
iris = datasets.load_iris()
X = iris['data']
Y = iris['target']
```

```python
# 划分训练集和测试集
index = np.arange(150)
np.random.shuffle(index)
# print(index)
x_train,x_test = X[index[:100]],X[index[100:]]
y_train,y_test = Y[index[:100]],Y[index[-50:]]
```

```python
# 建立模型
knn = KNeighborsClassifier(n_neighbors=5,weights='distance',algorithm='auto',p=1,n_jobs=-1)
```

```python
knn.fit(x_train,y_train)
y_ = knn.predict(x_test)
# 返回各个类别的概率
y2_ = knn.predict_proba(x_test)
# 返回最大值的索引
y2_.argmax(axis=1)


print(knn.score(x_test,y_test))
# print(y_test)
# print('------')
# print(y_)

print((y_ == y_test).sum()/50)
```
