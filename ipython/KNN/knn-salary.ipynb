```python
data = pd.read_csv('../datas/adult.data',names=['age','workclass','final_weight','education','education-num','marital-status',
                                              'occupation','relationship','race','sex','capital-gain','capital-loss',
                                               'hours-per-week','native-country','salary'])

```

```python
data.drop(labels=['final_weight','education','capital-gain','capital-loss'],axis=1,inplace=True)
```

```python
X = data.iloc[:,0:-1]
y = data['salary']
```

```python
##String串转换成int 或float，使算法可以计算
# map apply transform
```

```python
u = X['workclass'].unique()
# X.groupby('workclass')['age'].count()
# X[X.workclass == ' Without-pay']
```

```python
def convert(x):
    return np.argwhere(u == x)[0,0]
    
X['workclass'] = X['workclass'].map(convert)
```

```python
X.head()
```

```python
X.columns
cols = ['marital-status', 'occupation',
       'relationship', 'race', 'sex', 'native-country']
for col in cols:
    u = X[col].unique()
    def convert(x):
        return np.argwhere(u == x)[0,0]
    X[col] = X[col].map(convert)
X.head()
```

```python
knn = KNeighborsClassifier()
kfold = KFold(10)
accuracy = 0
for train,test in kfold.split(X,y):
    knn.fit(X.loc[train],y[train])
    acc = knn.score(X.loc[test],y[test])
    accuracy += acc/10
    
print(accuracy)


```
