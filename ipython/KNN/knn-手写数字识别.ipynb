```python
X = []
for i in range(10):
    for j in range(1,501):
        digit = cv2.imread('./datas/%d/%d_%d.bmp'%(i,i,j))
        X.append(digit[:,:,0])  #因为三个通道的值是一样的，所以可以任意取一个通道的颜色
        # cv2.cvtColor()
X = np.asarray(X)
y = np.array([i for i in range(10)] * 500)
y.sort()
```

```python
index = np.random.randint(0,5000,size=1)[0]
digit = X[index]
print(y[index])
plt.imshow(digit,cmap=plt.cm.grap)  #显示这张图片
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)  #分割成训练数据和测试数据
X_train = X_train.reshape(4000,-1)
X_test = X_test.reshape(4000,-1)
```

```python
knn = KNeighborsClassifier(n_jobs=63)
knn.fit(X_train,y_train)
y_ = knn.predict(X_test)
(y_ == y_test).sum()/1000
```
