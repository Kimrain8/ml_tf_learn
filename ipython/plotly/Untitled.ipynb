```python
#特征值和特征向量及矩阵的分解。
import numpy as np
import pandas as pd
from scipy import  linalg
from sympy import *

import matplotlib.pyplot as plt

import plotly.io as pio
import plotly.graph_objects as go
import plotly.express as px
from IPython.display import HTML


pio.renderers.default = 'iframe' # or 'notebook' or 'colab' or 'jupyterlab'

def my_project(a,b): #a向b向量投影
    return (b * a@b)/(np.linalg.norm(b)**2)

def SSP(a): #向量投影空间
    if a.ndim == 1:
        a = a.reshape(-1,1)
    return (a@a.T)/(a.T@a)

def PM(a): #矩阵投影空间
    return a@np.linalg.inv(a.T@a)@a.T
```

```python
#查看A@A.T  特征向量的轨迹。
B = np.array([1,5]).reshape(-1,1)
a1 = np.array(np.arange(-3,3,0.1)) 
a2 = np.array(np.arange(-3,3,0.1))

a1,a2 = np.meshgrid(a1,a2)
a1 = a1.ravel()
a2 = a2.ravel()

cnt = 0

x1 = []
y1 = []
x2 = []
y2 = []
txt1 = []
txt2 = []
c1 = []
c2 = []
seq = []
fms = []

for i in range(0,a1.size):
    A = np.array([a1[i],a2[i]]).reshape(-1,1)
    m1 = np.hstack((A,B))
    m1t = m1.T
    m2 = m1@m1t

    ev1,ev2 = np.linalg.eig(m2)
#     if i<= 10:
#         print(m1)
#         print(m2)
#         print(ev2)
    if not(isinstance(ev2[0,0],complex) or isinstance(ev2[1,0],complex)) and not(isinstance(ev2[0,1],complex) or isinstance(ev2[1,1],complex)):
        x1.append(ev2[0,0])
        y1.append(ev2[1,0])
        txt1.append(str(m1))
#         c1.append((i)%10)
        c1.append(1)
        x2.append(ev2[0,1])
        y2.append(ev2[1,1])
        txt2.append(str(m1))
#         c2.append((i)%10)
        c2.append(2)
        seq.append(i)
        fms.append(go.Frame(data=[go.Scatter(x=[ev2[0,0], ev2[0,1]], y=[ev2[1,0], ev2[1,1]],mode='markers')]))
#     if not(isinstance(ev2[0,1],complex) or isinstance(ev2[1,1],complex)):
#         x2.append(ev2[0,1])
#         y2.append(ev2[1,1])
#         txt2.append(str(m1))
#         c2.append(2)
    else:
        if cnt < 5:
            print(m2)
            cnt += 1

df =pd.DataFrame({'x1':x1,'y1':y1,'txt1':txt1,'c1':c1,'x2':x2,'y2':y2,'c2':c2,'txt2':txt2,'seq':seq}) #,'a1':a1,'a2':a2

fig1 = px.scatter(df,x='x1',y='y1',hover_data=['txt1'],color='c1') #,animation_frame='seq',hover_data=['a1','a2']
fig2 = px.scatter(df,x='x2',y='y2',hover_data=['txt2'],color='c2') #,animation_group="gp"),animation_frame='seq',hover_data=['a1','a2']
# fig = go.Figure(data=fig1.data + fig2.data)

fig = go.Figure(data=fig1.data + fig2.data,
                layout=go.Layout(
        xaxis=dict(range=[-3, 3], autorange=True), #range=[0, 5], 
        yaxis=dict(range=[-3, 3], autorange=True), #range=[0, 5], 
        title="Start Title",
        updatemenus=[dict(
            type="buttons",
            buttons=[dict(label="Play",
                          method="animate",
                          args=[None])])]
    ),
               frames=fms)
fig.show()

# fig1 = go.Scatter(x=[ev2[0,0], ev2[0,1]], y=[ev2[1,0], ev2[1,1]],mode='markers')
# fig = go.Figure(fig1)
# fig.show()


```
