```python
class Dog:
    def bark(self):
        print( 'Woof!')
 
class Husky(Dog):
    def bark(self):
        print('Howl!')

a = Husky()
a.bark()
```

```python
def newbark(self):
    print('New Woof!')

def howl(self):
    print('New Howl!')

Husky.bark = newbark  #替换旧方法
Dog.howl = howl  #增加一个新方法

a.bark()
b = Husky()
b.bark()
b.howl()
```

```python
def herd(self):
    self.bark()
 
border_collie = Dog()
border_collie.bark()
border_collie.herd = herd
border_collie.herd()
```

```python
import types
def herd(self):
    self.bark()

border_collie = Dog()
border_collie.herd  = types.MethodType(herd, border_collie)
 
border_collie.herd()
 
# border_collie.herd(sheep)
```
