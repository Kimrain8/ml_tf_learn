```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

import matplotlib.pyplot as plt
%matplotlib inline
from sklearn import datasets
from sklearn.model_selection import train_test_split
plt.rcParams['font.sans-serif']=[u'SimHei']
plt.rcParams['axes.unicode_minus']=False
```

```python
boston = datasets.load_boston()
X = boston.data
y = boston.target
```

```python
X = np.linspace(0,10,50).reshape(-1,1)
y = np.random.randint(2,8,size=1)*X

```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.coef_
```

$$\hat{w} = (X^TX)^{-1}X^Ty$$

```python
# 线性代数中的矩阵运算
np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)

lr = LinearRegression(fit_intercept=True)
lr.fit(X_train,y_train)
display(lr.coef_,lr.intercept_)

y_ = lr.predict(X_train)

```

```python
np.linspace(np.min(np.concatenate((y_train,y_))),np.max(np.concatenate((y_train,y_))),404)
# type(y_)
```

```python
plt.plot(np.linspace(np.min(np.concatenate((y_train,y_))),np.max(np.concatenate((y_train,y_))),404),y_train,'ro',label='真实值')
plt.plot(np.linspace(np.min(np.concatenate((y_train,y_))),np.max(np.concatenate((y_train,y_))),404),y_,'g*',label='预测值')
plt.legend(loc='upper left')
```
