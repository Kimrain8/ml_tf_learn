```python
#使用梯度下降求解最小值
d = lambda x : 2*(x-3) + 2.5
learning_rate = 0.1
```

```python
min_value = np.random.randint(-3,5,size=1)[0]

print('----------',min_value)
min_value_last = min_value + 0.1  #记录与前一个值的差值如果很小了，就表示完成

#tollerence
tol = 0.0001
count = 0
while True:
    if np.abs(min_value - min_value_last) < tol :
        break
    min_value_last = min_value
    min_value = min_value - learning_rate*d(min_value)
    count += 1
    print('+++++++++%d'%(count),min_value)
print('*********',min_value)
```

```python
d2 = lambda x : -2*(x-3) + 2.5
learning_rate = 0.1

max_value = np.random.randint(0,8,size=1)[0]

print('----------',max_value)
max_value_last = max_value + 0.01  #记录与前一个值的差值如果很小了，就表示完成

#tollerence
tol = 0.0001
count = 0
while True:
    if np.abs(max_value - max_value_last) < tol :
        break
    max_value_last = max_value
    max_value = max_value + learning_rate*d2(max_value)
    count += 1
    print('+++++++++%d'%(count),max_value)
print('*********',max_value)
```
