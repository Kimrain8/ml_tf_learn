由余弦公式：$d^2 = r^2 + r^2 - 2*r*r*\cos\alpha$  
由弧长公式：$\alpha * r = l$

```python
from scipy.optimize import fsolve
import numpy as np
from scipy.optimize import root


from sympy import *

```

```python
#  cos(x1) - 1 + 140^2 / 2*(x2)^2     1 - (x2)*(x1)
def f(x):
    d = 140
    l = 156
    a = float(x[0])
    r = float(x[1])
    return [np.cos(a) - 1 + (d**2)/(2*r**2),l - r * a]
res = fsolve(f,[1,1])
res
```
