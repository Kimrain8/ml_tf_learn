```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler,StandardScaler
from sklearn.linear_model import LassoCV
import matplotlib.pyplot as plt
import matplotlib as mpl
import os
%matplotlib inline
```

```python
if __name__ == "__main__":
    mpl.rcParams['font.sans-serif'] = [u'simHei']
    mpl.rcParams['axes.unicode_minus'] = False
    
    data = pd.read_excel('data2.xlsx',sheet_name='Sheet1',header=0)
    data.head()
    data.columns
    columns = [c for c in data.columns]
    data = data.values
    x = data[:,2:].astype(np.float)
    y = data[:,1].astype(np.int)
    
    ss = StandardScaler()
    x = ss.fit_transform(x)
    
    model = LassoCV(alphas=np.logspace(-3,2,10))
    model.fit(x,y)
    y_ = model.predict(x)
    display(model.alphas)
    display(model.coef_)
    display(model.score(x,y))
    display(y_)
    display(y)
    
    t = np.arange(len(y))
    plt.plot(t,y,'r-',ms=10,label=u'实际值')
    plt.plot(t,y_,'g-',ms=8,label=u'预测值')
    plt.grid(True)
    plt.show()
    
    
    
    
```
