```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,10,0.1)
y1=np.sin(x) + 2*np.cos(x)
y2 = 2*np.sin(x) + np.cos(x)
```

```python
x
```

```python
#实现双轴案例
# 创建图表和子图对象
fig, ax1 = plt.subplots()

# 绘制第一条曲线在ax1上，并设置Y轴标签
ax1.plot(x, y1, 'b-')
ax1.set_xlabel('X data')
ax1.set_ylabel('Y values for exp(-x)', color='b')

# 创建第二个Y轴
ax2 = ax1.twinx()
ax2.plot(x, y2, 'r-')
ax2.set_ylabel('Y values for ln(x)', color='r')

# 显示图表
plt.show()
```
