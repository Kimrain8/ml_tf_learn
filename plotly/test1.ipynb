```python
import plotly.express as px
import plotly.graph_objects as go

# 创建基础散点图矩阵
iris = px.data.iris()
fig = px.scatter_matrix(iris, dimensions=iris.columns[:4])

# 获取维度数量
n_dims = len(iris.columns[:4])

# 移除或替换对角线上的域类型子图
for i in range(1, n_dims + 1):
    # 1. 移除现有轨迹
    fig.data = [trace for trace in fig.data if not (
        trace.type == 'histogram' and 
        f"x{i}" in trace.xaxis and 
        f"y{i}" in trace.yaxis
    )]
    
    # 2. 添加新的笛卡尔直方图
    col_name = iris.columns[i-1]
    fig.add_trace(
        go.Histogram(
            x=iris[col_name],
            name=f"{col_name}分布",
            marker_color='#1f77b4'
        ),
        row=i, col=i
    )
    
    # 3. 更新子图类型为笛卡尔
    fig.update_layout({
        f"xaxis{i}": dict(anchor=f"y{i}", type="linear"),
        f"yaxis{i}": dict(anchor=f"x{i}", type="linear")
    })

fig.show()
```
